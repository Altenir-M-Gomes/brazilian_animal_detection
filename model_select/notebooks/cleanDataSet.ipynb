```python
from PIL import Image, UnidentifiedImageError
import os

def listar_imagens_corrompidas(caminho_pasta):
    imagens_corrompidas = []

    for nome_arquivo in os.listdir(caminho_pasta):
        
        caminho_imagem = os.path.join(caminho_pasta, nome_arquivo)
        try:
            with Image.open(caminho_imagem) as img:
                img.load()  # força leitura completa da imagem
                img.convert("RGB")  # se der erro aqui, imagem está com problema de leitura/conversão

        except (OSError, UnidentifiedImageError):
            imagens_corrompidas.append(caminho_imagem)

    print(f"\nTotal de imagens corrompidas: {len(imagens_corrompidas)}")
    for img_path in imagens_corrompidas:
        print(f"- {img_path}")
    
    return imagens_corrompidas
```

```python
listar_imagens_corrompidas("../imagens/test/com_animal")
```
